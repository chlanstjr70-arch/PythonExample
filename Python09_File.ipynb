```python
f = open("demofile.txt")
print(f.read())
```

```python
with open("demofile.txt") as f:
  print(f.read())
```

```python
f = open("demofile.txt")
print(f.readline())
f.close()
```

```python
with open("demofile.txt") as f:
    print(f.read(5))
```

```python
with open("demofile.txt") as f:
    print(f.readline())
```

```python
with open("demofile.txt") as f:
    print(f.readline())
    print(f.readline())
```

```python
with open("demofile.txt") as f:
    for x in f:
        print(x)
```

```python
with open("demofile.txt", "a") as f:
  f.write("Now the file has more content!")
with open("demofile.txt") as f:
  print(f.read())
```

```python
with open("demofile.txt", "w") as f:
  f.write("Woops! I have deleted the content!")
with open("demofile.txt") as f:
  print(f.read())
```
